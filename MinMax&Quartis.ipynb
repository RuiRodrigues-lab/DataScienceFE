```python
import pandas as pd
import numpy as np

# Criar o DataFrame
dados = [
    [59.03, 95.16, 48.19, 55.16, 5.27, 24.14, 59.12, 47.50, 21.17, 32.43, 49.50, 72.20, 13.05],
    [37.12, 10.21, 56.91, 12.17, 10.17, 60.14, 22.23, 18.17, 45.12, 82.17, 66.67, 14.50, 43.57],
    [62.17, 47.35, 27.53, 23.12, 75.17, 37.05, 9.25, 67.78, 36.18, 87.50, 62.31, 19.21, 58.14],
    [15.05, 28.62, 52.19, 78.75, 38.92, 35.64, 90.12, 25.14, 95.62, 73.14, 25.00, 32.19, 42.71]
]

df = pd.DataFrame(dados)

# Flatten dos valores
valores = df.values.flatten()

# Definir intervalos (equivalente a seq(5, 100, l=6))
bins = np.linspace(5, 100, 6)  # cria 5 intervalos entre 5 e 100

# Criar categorias (equivalente ao cut() em R), aqui cortamos os "valores" e separamos nos respectivos "bins"
categorias = pd.cut(valores, bins=bins, include_lowest=True)

# Tabela de frequências absolutas
freq_abs = categorias.value_counts().sort_index()

# Tabela de frequências relativas (%)
freq_rel = (freq_abs / freq_abs.sum()).sort_index() * 100

# Juntar tudo num DataFrame
tabela_frequencias = pd.DataFrame({
    "Intervalo": freq_abs.index.astype(str),
    "Frequência": freq_abs.values,
    "Percentagem": freq_rel.round(2).values
})

print(tabela_frequencias)
```
