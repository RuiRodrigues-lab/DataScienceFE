```python
#importar a lib como pd e np
import pandas as pd
import numpy as np
```

```python
#Aqui definimos os dados da amostra test
dados = {
    "Grau_Escolaridade": (
        ["Básico"] * 3 +
        ["Secundário"] * 5 +
        ["Licenciatura"] * 13 +
        ["Mestrado"] * 4
    )
}

df = pd.DataFrame(dados)
df.head()
```

```python
#Frequencias Absolutas
freq_abs = df["Grau_Escolaridade"].value_counts()
freq_abs
```

```python
#Frequencias Relativas
freq_rel = df["Grau_Escolaridade"].value_counts(normalize=True).round(2)
freq_rel
```

```python
#Aqui juntamos ambas as columas que calculamos antes em uma tabela
tabela = pd.DataFrame({
    "Frequência": freq_abs,
    "Frequência Relativa": freq_rel
})
tabela
```

```python
# Moda (pode devolver várias em caso de empate)
moda = df["Grau_Escolaridade"].mode()
print("Moda:", list(moda))

# Frequência da(s) moda(s) e percentagem
vc = df["Grau_Escolaridade"].value_counts()
freq_max = vc.max()
modas = vc[vc == freq_max]
print(" ")
print("Frequências da(s) moda(s):")
print(modas)
print(" ")
print("Percentagem:", freq_max / len(df))
```
