```python
n_total = len(df["Antigo"])
n_non_missing = df["Antigo"].notna().sum()

print("Total Antigo:", n_total)
print("Não-missing:", n_non_missing)

n_total = len(df["Atual"])
n_non_missing = df["Atual"].notna().sum()

print("\nTotal Atual:", n_total)
print("Não-missing:", n_non_missing)
```

```python
# Teste de normalidade (Shapiro-Wilk) (Para testar a normalidade da amostra!)
shapiro_result = stats.shapiro(df["Antigo"].dropna())
print(shapiro_result)

# Teste de normalidade (Shapiro-Wilk) (Para testar a normalidade da amostra!)
shapiro_result = stats.shapiro(df["Atual"].dropna())
print(shapiro_result)

#se p>0.05, assumimos normalidade, rejeitamos h0
```

```python
# Dados
a = df["Antigo"].dropna()
b = df["Atual"].dropna()

# Paired t-test (equivalente ao R)
t_stat, p_value = ttest_rel(a, b, alternative="less")

print("Paired t-test")
print(f"t = {t_stat}, df = {len(a)-1}, p-value = {p_value}")
print("alternative hypothesis: true mean difference is less than 0")

# Médias
mean_antigo = a.mean()
mean_atual = b.mean()

# Desvios-padrão (amostrais, iguais ao R)
sd_antigo = a.std(ddof=1)
sd_atual = b.std(ddof=1)

print("\nmean(CP4$Antigo) =", mean_antigo)
print("mean(CP4$Atual)  =", mean_atual)

print("\nsd(CP4$Antigo)   =", sd_antigo)
print("sd(CP4$Atual)    =", sd_atual)

#Participaram no estudo 20 utilizadores. Em média, a satisfação dos participantes no estudo com o site
#antigo (68,45 pontos) é inferior à do site atual (77,15 pontos). A dispersão da satisfação (relativamente
#à média) com o site antigo (2,910507 pontos) é inferior à do site atual (4,901933 pontos).
#p-value=2,566x10-80 → rejeitar H0
#Para qualquer nível de significância (0,01 ou 0,05 ou 0,1), existe evidência estatística para rejeitar H0,
#a satisfação média dos utilizadores com o site antigo é significativamente inferior à satisfação média
#dos utilizadores com o site atual. Os utilizadores estão mais satisfeitos com o site atual.
```
