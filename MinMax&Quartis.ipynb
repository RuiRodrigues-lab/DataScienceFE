```python
import pandas as pd

# Criar o DataFrame
dados = [
    [59.03, 95.16, 48.19, 55.16, 5.27, 24.14, 59.12, 47.50, 21.17, 32.43, 49.50, 72.20, 13.05],
    [37.12, 10.21, 56.91, 12.17, 10.17, 60.14, 22.23, 18.17, 45.12, 82.17, 66.67, 14.50, 43.57],
    [62.17, 47.35, 27.53, 23.12, 75.17, 37.05, 9.25, 67.78, 36.18, 87.50, 62.31, 19.21, 58.14],
    [15.05, 28.62, 52.19, 78.75, 38.92, 35.64, 90.12, 25.14, 95.62, 73.14, 25.00, 32.19, 42.71]
]

df = pd.DataFrame(dados)

# Converter todos os valores para uma lista plana
valores = df.values.flatten()

# Criar resumo global
resumo_global = pd.DataFrame({
    "estatistica": ["min", "max", "count", "média", "mediana", "desvio_padrao"],
    "valor": [
        valores.min(),
        valores.max(),
        len(valores),
        valores.mean(),
        pd.Series(valores).median(),
        valores.std()
    ]
})

print(resumo_global)

```
