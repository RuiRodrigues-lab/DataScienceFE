```python
import pandas as pd
from pandas.api.types import CategoricalDtype
import numpy as np
import matplotlib.pyplot as plt
#Precisamos desta biblioteca para podermos escolher um ficheiro local
#Se o ficheiro vier por API ou tivermos um link, é so alterar a forma de import
from google.colab import files

# 1️⃣ Faz upload do ficheiro (vai abrir uma janela para escolher no teu PC)
uploaded = files.upload()

# 2️⃣ Guarda o nome do ficheiro (Colab mostra o nome depois do upload)
filename = list(uploaded.keys())[0]

# 3️⃣ Lê o Excel, por default lê sempre a primeira tab, por isso podemos usar o "sheet_name"
# Se tivermos dados em varias tabs, devemos usar uma Dataframe(df) para cada uma das tabs
df = pd.read_excel(filename, sheet_name='Exerc2')
df.head()
```

```python
df.info()
```

```python
# 4️⃣ Converter para factores (ambas qualitativas nominais, por isso não precisam ser ordenadas)
df['Sexo'] = df['Sexo'].astype('category')
df['Bebida'] = df['Bebida'].astype('category')

# 5️⃣ Tabelas de contingência
tab = pd.crosstab(df['Bebida'], df['Sexo'])
print("\n📊 Tabela de Contingência:")
print(tab)
```

```python
# 6️⃣ Proporções (% do total)
prop_total = pd.crosstab(df['Bebida'], df['Sexo'], normalize='all') * 100
print("\n📊 Proporções (%) sobre o total:")
print(prop_total.round(1))
```

```python
# 7️⃣ Tabela inversa (Sexo por Escalão)
tab_inv = pd.crosstab(df['Sexo'], df['Bebida'])
print("\n📊 Tabela inversa:")
print(tab_inv)

prop_total_inv = pd.crosstab(df['Sexo'], df['Bebida'], normalize='all') * 100
print("\n📊 Proporções (%) (inversa):")
print(prop_total_inv.round(1))
```

```python
# --------------------------------------------------
# 5️⃣ Gráficos
# --------------------------------------------------
plt.style.use('seaborn-v0_8-whitegrid')
cores = ["yellow", "blue"]
cores_inv = ["green", "pink", "orange"]

# --------------------------------------------------
# G1 — Barras lado a lado
# --------------------------------------------------
ax = tab.plot(kind='bar', color=cores)
plt.title("Gráfico de Barras")
plt.ylabel("Frequência Absoluta")
plt.xlabel("")
plt.ylim(0, 60)
plt.legend(title="Sexo", loc='upper right')

for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', fontsize=9)
plt.show()
```

```python
# --------------------------------------------------
# G2 — Barras lado a lado
# --------------------------------------------------
ax = prop_total.plot(kind='bar', color=cores)
plt.title("Gráfico de Barras")
plt.ylabel("Percentagens (%)")
plt.xlabel("")
plt.ylim(0, 30)
plt.legend(title="Sexo", loc='upper right')

for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', fontsize=9)
plt.show()
```

```python
# --------------------------------------------------
# G3 — Barras lado a lado
# --------------------------------------------------
ax = tab_inv.plot(kind='bar', color=cores_inv)
plt.title("Gráfico de Barras")
plt.ylabel("Frequencias Absolutas")
plt.xlabel("")
plt.ylim(0, 50)
plt.legend(title="Bebida", loc='upper right')

for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', fontsize=9)
plt.show()
```

```python
# --------------------------------------------------
# G4 — Barras lado a lado
# --------------------------------------------------
ax = prop_total_inv.plot(kind='bar', color=cores_inv)
plt.title("Gráfico de Barras")
plt.ylabel("Percentagens (%)")
plt.xlabel("")
plt.ylim(0, 40)
plt.legend(title="Bebida", loc='upper right')

for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', fontsize=9)
plt.show()
```
